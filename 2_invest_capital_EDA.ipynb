```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
invest_cap = pd.read_csv("csv sdmx_data_1561.csv")
invest_cap.head()
```

```python
invest_cap.info()
```

```python
invest_cap.describe()
```

```python
corr_matrix = invest_cap.select_dtypes(include="number").corr()
corr_matrix
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(corr_matrix, cmap="coolwarm", annot=False)
plt.title("Корреляции между квартальными показателями инвестиций")
plt.show()
```

Преобразуем таблицу в удобный формат для визуализации (код ниже)

```python
invest_cap_melted = invest_cap.melt(
    id_vars=invest_cap.columns[:5],   # первые 5 — это не временные
    var_name="Period",
    value_name="Investment"
)
invest_cap_melted.head()
```

```python
invest_cap_melted  = invest_cap_melted[ ['Code', 'Klassifikator_ru', 'Klassifikator_en', 'Period', 'Investment'] ]
```

```python
invest_cap_melted.head()
```

Построение общей динамики - мы сначала агрегируем данные, чтобы получить общие значения по всем категориям (если в данных несколько кодов),
а потом построим линию, показывающую, как менялся общий объём инвестиций с 2020 по 2025 год.

```python
# агрегируем данные по кварталам
invest_cap_total = invest_cap_melted.groupby("Period", as_index=False)["Investment"].sum()

# строим график
plt.figure(figsize=(10,5))
sns.lineplot(data=invest_cap_total, x="Period", y="Investment", marker="o")
plt.title("Динамика общего объема инвестиций в основной капитал (2020–2025 гг.)")
plt.xlabel("Период")
plt.ylabel("Объем инвестиций, млрд сум")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

Сравнение направлений (в нашем случае — регионов)

```python
plt.figure(figsize=(12,6))
sns.lineplot(
    data=invest_cap_melted,
    x="Period", 
    y="Investment", 
    hue="Klassifikator_ru", 
    marker="o"
)
plt.title("Динамика инвестиций в основной капитал по регионам (2020–2025 гг.)")
plt.xlabel("Период")
plt.ylabel("Объем инвестиций, млрд сум")
plt.xticks(rotation=45)
plt.legend(title="Регион", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
invest_gdp = pd.read_csv("C:\Нурик\универ\диплом\данные\инвестиции\Доля инвестиций в основной капитал в ВВП\sdmx_data_1340.csv")
invest_gdp.head()
```

```python
Преобразование данных
```

```python
gdp_melted = invest_gdp.melt(
    id_vars=invest_gdp.columns[:5],
    var_name="Year",
    value_name="GDP_share"
)

gdp_melted
```

```python
gdp_melted = gdp_melted[ [ 'Code', 'Klassifikator_ru', 'Klassifikator_en', 'Year', 'GDP_share' ] ]
gdp_melted.head()
```

```python
Построение графика
```

```python
plt.figure(figsize=(10, 6))
plt.plot(gdp_melted['Year'], gdp_melted['GDP_share'])
plt.xticks(gdp_melted['Year'][::2], rotation=45)
plt.title("Доля инвестиций в ВВП по годам")
plt.xlabel("Год")
plt.ylabel("Доля инвестиций в ВВП, %")
plt.grid(True)
plt.show()
```

Поиск зависимостей между показателями


```python
# извлекаем год из 'Period' (если формат 2020Q1 или 2020-Q1)
invest_cap_total["Year"] = invest_cap_total["Period"].astype(str).str[:4].astype(int)

# приводим 'Year' к int и убираем дубликаты
gdp_melted["Year"] = gdp_melted["Year"].astype(int)

# объединяем по году
merged = pd.merge(invest_cap_total, gdp_melted, on="Year", how="inner")

merged.head()
```

```python
Визуализация зависимости между объёмом инвестиций и долей инвестиций в ВВП
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=merged, x="Investment", y="GDP_share")
sns.regplot(data=merged, x="Investment", y="GDP_share", scatter=False, color="red")
plt.title("Зависимость между объёмом инвестиций и долей инвестиций в ВВП")
plt.xlabel("Объём инвестиций, млрд сум")
plt.ylabel("Доля инвестиций в ВВП, %")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
Расчёт коэффициента корреляции между объёмом инвестиций и долей инвестиций в ВВП
```

```python
from scipy.stats import pearsonr

# вычисляем коэффициент корреляции Пирсона
corr, p_value = pearsonr(merged["Investment"], merged["GDP_share"])

print("Коэффициент корреляции:", round(corr, 3))
print("p-value:", round(p_value, 5))
```
